```python
#verifica se todos os links estão funcionando (resposta deve ser um application/pdf pra cada município)
import requests
from concurrent.futures import ThreadPoolExecutor

def verificar_url(url):
    municipio = url.split("/")[-1].replace(".pdf", "")

    try:
        response = requests.get(url, timeout=30)

        content_type = response.headers.get("Content-Type", "")

        if not response.ok:
            return municipio, False, f"HTTP {response.status_code}"

        if "application/pdf" not in content_type.lower():
            return municipio, False, f"Content-Type: {content_type}"

        return municipio, True, content_type

    except requests.RequestException as e:
        return municipio, False, str(e)


falhas = 0



with ThreadPoolExecutor(max_workers=10) as executor:
    resultados = executor.map(verificar_url, urls)

    for municipio, sucesso, mensagem in resultados:
        if sucesso:
            print(f"{municipio} - {mensagem}")
        else:
            print(f"{municipio} - FALHA - {mensagem}")
            falhas += 1

print(f"Houve {falhas} falhas")
```

```python
#baixa os pdfs
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed
import requests



DESTINO = Path("archives")
DESTINO.mkdir(exist_ok=True)

def baixar_pdf(url):
    nome_arquivo = url.split("/")[-1]
    caminho = DESTINO / nome_arquivo

    if caminho.exists():
        return f"Já existe: {nome_arquivo}"

    try:
        resp = requests.get(url, timeout=60, stream=True)
        resp.raise_for_status()

        with open(caminho, "wb") as f:
            for chunk in resp.iter_content(chunk_size=8192):
                if chunk:
                    f.write(chunk)

        return f"OK: {nome_arquivo}"

    except Exception as e:
        return f"ERRO: {nome_arquivo} -> {e}"

# Baixa vários PDFs em paralelo
with ThreadPoolExecutor(max_workers=10) as executor:
    futures = [executor.submit(baixar_pdf, url) for url in urls]

    for future in as_completed(futures):
        print(future.result())

print(f"\nConcluído. PDFs salvos em: {DESTINO.resolve()}")
```

```python
from pathlib import Path
import camelot
import os

Path("municipios").mkdir(exist_ok=True)

archives = [
    arq for arq in sorted(os.listdir("archives"))
    if arq.endswith(".pdf")
]

print(archives)

def processar(arch):
    name = Path(arch).stem
    Path("municipios", name).mkdir(parents=True, exist_ok=True)

    tables = camelot.read_pdf(
        f"archives/{arch}",
        pages="all",
    )

    tables.export(
        f"municipios/{name}/{name}.csv",
        f="csv",
        compress=False,
    )
    return name

```

```python
from concurrent.futures import ProcessPoolExecutor
from tqdm.auto import tqdm

with ProcessPoolExecutor() as executor:

    resultados = executor.map(processar, archives)

    for municipio in tqdm(
        resultados,
        total=len(archives),
        desc="Descompactando PDFs"
    ):
        tqdm.write(f"✓ {municipio} ok")
```
